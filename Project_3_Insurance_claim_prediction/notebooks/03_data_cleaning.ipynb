```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120) 
```

```python
#Load Train data
train_path = '../data/Train_data.csv'
train_df = pd.read_csv(train_path)
```

```python
#Separate target variable and features
target = 'Claim'
X = train_df.drop(columns=[target])
y = train_df[target]
```

```python
#Handling Duplicate Rows
duplicate_rows = train_df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_rows}")

if duplicate_rows > 0:
    X = X.drop_duplicates() # Drop duplicate rows from features
    y = y.loc[X.index] # Align target variable with updated features

print(f"Shape of features after removing duplicates: {X.shape}")
print(f"Shape of target variable after removing duplicates: {y.shape}")
```

```python
#Identifying numerical and categorical columns
numerical_cols = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()
```

Now that the data is separate in Numerical and Cathegorical we will choose a technique for handling the missing values for each of them:
* *Numerical columns: We will use the median Technique. By replacing all the missing values by the median value of the column*
* *Categorical columns: We will replace all the missing values by the string "Unknown"*

```python
#Missing Values Treatment for numerical columns
for col in numerical_cols:
    if X[col].isnull().sum() > 0:
        X[col] = X[col].fillna(X[col].median())
```

```python
#Missing Values Treatment for categorical columns
for col in categorical_cols:
    if X[col].isnull().sum() > 0:
        X[col] = X[col].fillna('Unknown')
```

```python
#Remove Identifier Columns that is Customer ID
if 'CustomerID' in X.columns:
    X = X.drop(columns=['CustomerID'])
```

```python
#Saving cleaned data
cleaned_df = X.copy()
cleaned_df[target] = y
cleaned_df.to_csv('../data/cleaned_train_data.csv', index=False)
```
