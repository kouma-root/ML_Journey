```python
#Libraries importation
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
#Data loading
data_path = '../data/cleaned_train_data.csv'
df = pd.read_csv(data_path)
```

```python
#Separate features and target variable
target = 'Claim'
X = df.drop(columns=[target])
y = df[target]
```

```python
#Separate Numerical and Categorical Columns
numerical_cols = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()
```

### Feature Engineering: 
We will log-transform highly skewed numerical features

```python
#Feature Engineering:
skewed_cols = X[numerical_cols].skew().abs()
skewed_cols = skewed_cols[skewed_cols > 1].index.tolist()

for col in skewed_cols:
    X[col] = np.log1p(X[col])
    
print("Log transformation features:", skewed_cols)
```

One-hot Encode Categorical Variables

```python
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
```

```python
#Train-Validation Split
X_train, X_val, y_train, y_val = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y)
```

```python
#Scaling Numerical Features
scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_val_scaled = X_val.copy()

X_train_scaled[numerical_cols] = scaler.fit_transform(X_train_scaled[numerical_cols])
X_val_scaled[numerical_cols] = scaler.transform(X_val_scaled[numerical_cols])
```

```python
#Save processed data
X_train_scaled.to_csv('../data/X_train_processed.csv', index=False)
X_val_scaled.to_csv('../data/X_val_processed.csv', index=False)
y_train.to_csv('../data/y_train.csv', index=False)
y_val.to_csv('../data/y_val.csv', index=False)
```
